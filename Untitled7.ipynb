```python
total_sum = 0
for i in range(10):
    number = int(input("Enter a number: "))
    if number == 0:
        break
    if number < 0:
        continue
    if number % 2 == 0:
        total_sum += number
print("Final sum of even positive numbers:", total_sum)
```

```python
number = int(input("Enter a number: "))

for i in range(1, 13):
    product = number * i

    if product % 3 == 0:
        continue

    if product > 50:
        break

    print(f"{number} x {i} = {product}")
```

```python
secret_number = 15
attempts = 0

while True:
    guess = int(input("Guess a number between 1 and 20: "))

    if guess < 1 or guess > 20:
        print("Invalid guess. Try again.")
        continue

    attempts += 1

    if guess == secret_number:
        print("Correct! 🎉")
        break
    else:
        print("Wrong guess. Try again.")

print("Valid attempts:", attempts)
```

```python
n = int(input("Enter a number: "))
count = 0

for i in range(1, n + 1):

    if i > 50:
        break

    if i % 8 == 0:
        continue

    if i % 4 == 0:
        count += 1

print("Count:", count)
```
